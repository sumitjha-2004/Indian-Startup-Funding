```python
import pandas as pd 
import numpy as np 
```

```python
df = pd.read_csv('D:\coding\Indian Startup Funding\indian_startup_funding_2020_2025_sample.csv')
df.head()
```

```python
df.info()
```

```python
sorted(df['Investors'].unique().tolist())
```

```python
df['Investors'] = df['Investors'].str.split(',')
```

```python
df.rename(columns = {'InvestmentAmount_USD' : 'amount'}, inplace = True)
```

```python
df
```

```python
def usdtoinr(amount):
    return (amount*90)/10000000
```

```python
df['amount'] = usdtoinr(df['amount'])
```

```python
df.sample(5)
```

```python
df.info()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df.to_csv('startup_clean.csv', index = False)
```
